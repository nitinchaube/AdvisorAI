```python
def json_to_chunks(docs):
    all_chunks = []
    example_key = next(iter(docs))
    record_ids = docs[example_key].keys()  

    for record_id in record_ids:
        chunk = {}
        for key in docs:
            # Skip if the key is duplicated like 'Name', 'Name', etc.
            if key not in chunk:  
                chunk[key] = docs[key].get(record_id)
        all_chunks.append(chunk)

    return all_chunks

```

```python
def load_and_merge_json(path):
    all_chunks = []
    for filename in os.listdir(path):
        if filename.endswith('.json'):
            with open(os.path.join(path, filename),'r', encoding='utf-8') as f:
                data = json.load(f)
                all_chunks.extend(json_to_chunks(data))
                break
    return all_chunks
```

```python
from nltk.tokenize import word_tokenize
def get_max_token_length(chunks):
    max_length = 0
    longest_chunk = None
    for chunk in chunks:
        combined_text = " ".join(str(value) for value in chunk.values() if value)
        tokens = word_tokenize(combined_text)
        length = len(tokens)
        if length > max_length:
            max_length = length
            longest_chunk = chunk
    return max_length, longest_chunk
```

```python
from collections import deque
```

```python
class Solution:
    def __init__(self):
        self.coordinates = [
            [-1, -2], [-1, 2], [1, 2], [1, -2],
            [-2, -1], [-2, 1], [2, -1], [2, 1]
        ]
    def dfs_helper(self, src_x, src_y, tar_x, tar_y, n):
        visited =[[0 for _ in range(n)] for _ in range(n)]
        q = deque()
        q.append((src_x, src_y))
        visited[src_x][src_y] = 1
        steps = 0
        while q:
            q_size = len(q)
            while q_size>0:
                x, y = q.popleft()
                if x==tar_x and y==tar_y:
                    return steps
                
                for x1, y1 in self.coordinates:
                    new_x, new_y = x+x1, y+y1
                    if 0<=new_x<n and 0<=new_y<n and visited[new_x][new_y] == 0:
                        visited[new_x][new_y] = 1
                        q.append((new_x, new_y))

                q_size-=1
            steps+=1


        return -1



    def MinimumKnightMoves(self, source, target, n):
        src_x = n - source[1]
        src_y = source[0] - 1
        tar_x = n - target[1]
        tar_y = target[0] - 1
        return self.dfs_helper(src_x, src_y, tar_x, tar_y, n)

```

```python
finder = Solution()
KnightPos = [4, 5]
TargetPos = [1, 1]
n = 6
print(finder.MinimumKnightMoves(KnightPos, TargetPos, n))
```
